# Algoritmos de Ordenamiento

## ¿Qué es un algoritmo de ordenamiento?

Un algoritmo de ordenamiento es un procedimiento que **reorganiza** los elementos de una colecci\u00f3n (arreglo, lista, etc.) en un orden espec\u00edfico, generalmente de **menor a mayor** (orden ascendente) o de **mayor a menor** (orden descendente).

Dado un arreglo $A = [a_0, a_1, \dots, a_{n-1}]$, el objetivo es producir una permutaci\u00f3n $A' = [a'_0, a'_1, \dots, a'_{n-1}]$ tal que:

$$
a'_0 \leq a'_1 \leq \dots \leq a'_{n-1}
$$

Los algoritmos de ordenamiento son fundamentales en ciencias de la computaci\u00f3n. Muchas operaciones (como la b\u00fasqueda binaria) requieren datos ordenados, y el ordenamiento es una de las operaciones m\u00e1s estudiadas en la historia de la computaci\u00f3n.

### Conceptos importantes

- **In-place**: El algoritmo ordena el arreglo sin usar memoria adicional significativa (solo $O(1)$ extra).
- **Estable**: Si dos elementos tienen el mismo valor, mantienen su orden relativo original despu\u00e9s de ordenar.
- **Complejidad temporal**: Cu\u00e1ntas operaciones (comparaciones, intercambios) se necesitan.

Veremos cinco algoritmos cl\u00e1sicos:

| Algoritmo | Complejidad (peor) | In-place | Estable |
|---|---|---|---|
| **Bubble Sort** | $O(n^2)$ | S\u00ed | S\u00ed |
| **Selection Sort** | $O(n^2)$ | S\u00ed | No |
| **Insertion Sort** | $O(n^2)$ | S\u00ed | S\u00ed |
| **Merge Sort** | $O(n \log n)$ | No | S\u00ed |
| **Quick Sort** | $O(n^2)$ | S\u00ed | No |

**Usaremos el siguiente arreglo de ejemplo a lo largo del notebook:**

$$[38, 27, 43, 3, 9, 82, 10]$$

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

import warnings
warnings.filterwarnings("ignore")

arr = np.array([38, 27, 43, 3, 9, 82, 10])

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(range(len(arr)), arr, color='steelblue', edgecolor='black')
for i, v in enumerate(arr):
    ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
ax.set_xlabel('Índice')
ax.set_ylabel('Valor')
ax.set_title('Arreglo desordenado')
ax.set_xticks(range(len(arr)))
ax.grid(True, axis='y')
plt.show()
````

## Ordenamiento Burbuja (*Bubble Sort*)

### ¿Cómo funciona?

El Bubble Sort es el algoritmo de ordenamiento más simple. Funciona comparando **pares adyacentes** de elementos y **intercambiándolos** si están en el orden incorrecto. Este proceso se repite hasta que el arreglo esté ordenado.

El nombre "burbuja" viene de que los elementos más grandes van "subiendo" (como burbujas) hacia el final del arreglo en cada pasada.

**Pseudocódigo:**

```
función bubble_sort(A):
    n = tamaño(A)
    para i desde 0 hasta n-2:
        para j desde 0 hasta n-2-i:
            si A[j] > A[j+1]:
                intercambiar(A[j], A[j+1])
```

En cada pasada $i$, el elemento más grande del subarreglo no ordenado "burbujea" hasta su posición final.

### Complejidad computacional

En cada pasada $i$ (de $0$ a $n-2$), se hacen $n-1-i$ comparaciones:

$$
\text{Total de comparaciones} = \sum_{i=0}^{n-2} (n-1-i) = (n-1) + (n-2) + \dots + 1 = \frac{n(n-1)}{2}
$$

Por lo tanto:

- **Mejor caso** (arreglo ya ordenado, con optimizaci\u00f3n): $O(n)$
- **Peor caso** (arreglo en orden inverso): $O(n^2)$
- **Caso promedio**: $O(n^2)$

$$
T(n) = O(n^2)
$$

### Ejemplo paso a paso

Ordenemos $[38, 27, 43, 3, 9, 82, 10]$:

**Pasada 1** (el mayor llega al final):
- $[38, 27] \rightarrow$ intercambiar $\rightarrow [27, 38, 43, 3, 9, 82, 10]$
- $[38, 43] \rightarrow$ ok
- $[43, 3] \rightarrow$ intercambiar $\rightarrow [27, 38, 3, 43, 9, 82, 10]$
- $[43, 9] \rightarrow$ intercambiar $\rightarrow [27, 38, 3, 9, 43, 82, 10]$
- $[43, 82] \rightarrow$ ok
- $[82, 10] \rightarrow$ intercambiar $\rightarrow [27, 38, 3, 9, 43, 10, \mathbf{82}]$

Después de la pasada 1, el **82** está en su posición final.

El proceso continúa hasta que todo el arreglo esté ordenado: $[3, 9, 10, 27, 38, 43, 82]$

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

arr_orig = [38, 27, 43, 3, 9, 82, 10]
arr = arr_orig.copy()
n = len(arr)

# Registrar estado después de cada pasada
states = [arr.copy()]
for i in range(n - 1):
    for j in range(n - 1 - i):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
    states.append(arr.copy())

# Mostrar primera y última pasada + resultado
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

titles = ['Arreglo original', 'Después de pasada 1', 'Resultado final']
data = [states[0], states[1], states[-1]]
for ax_idx, (ax, state, title) in enumerate(zip(axes, data, titles)):
    sorted_positions = set()
    if ax_idx == 1:
        sorted_positions = {n - 1}  # 82 en posición final
    elif ax_idx == 2:
        sorted_positions = set(range(n))

    colors = ['#2ECC71' if i in sorted_positions else 'steelblue' for i in range(n)]
    ax.bar(range(n), state, color=colors, edgecolor='black')
    for i, v in enumerate(state):
        ax.text(i, v + 2, str(v), ha='center', fontweight='bold', fontsize=9)
    ax.set_title(title)
    ax.set_xticks(range(n))
    ax.set_xlabel('Índice')
    ax.grid(True, axis='y')

axes[0].set_ylabel('Valor')
legend_elements = [
    Patch(facecolor='steelblue', edgecolor='black', label='Sin ordenar'),
    Patch(facecolor='#2ECC71', edgecolor='black', label='En posición final'),
]
fig.legend(handles=legend_elements, loc='lower center', ncol=2, fontsize=9)
plt.suptitle('Bubble Sort — Progreso', fontsize=13, fontweight='bold')
plt.tight_layout(rect=[0, 0.06, 1, 0.95])
plt.show()
````

### Animación interactiva

La animación muestra cada comparación e intercambio. Los pares comparados se muestran en **naranja**, y los elementos que ya están en su posición final se muestran en **verde**.

````python
def bubble_sort_steps(arr):
    """Registra cada paso del Bubble Sort."""
    arr = list(arr)
    n = len(arr)
    steps = [('initial', list(arr), -1, -1, set())]
    sorted_indices = set()

    for i in range(n - 1):
        for j in range(n - 1 - i):
            # Comparar
            steps.append(('compare', list(arr), j, j + 1, set(sorted_indices)))
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                steps.append(('swap', list(arr), j, j + 1, set(sorted_indices)))
        sorted_indices.add(n - 1 - i)

    sorted_indices.add(0)
    steps.append(('done', list(arr), -1, -1, set(range(n))))
    return steps


def animate_bubble_sort(arr):
    """Crea una animación interactiva del Bubble Sort."""
    steps = bubble_sort_steps(arr)
    n = len(arr)
    max_val = max(arr)

    fig, ax = plt.subplots(figsize=(10, 5))
    bar_data = steps[0][1]
    bars = ax.bar(range(n), bar_data, color='steelblue', edgecolor='black')
    texts = []
    for i, v in enumerate(bar_data):
        t = ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
        texts.append(t)

    ax.set_xlabel('Índice')
    ax.set_ylabel('Valor')
    ax.set_title('Bubble Sort')
    ax.set_xticks(range(n))
    ax.set_ylim(0, max_val * 1.25)
    ax.grid(True, axis='y')

    status = ax.text(0.5, 1.08, '', transform=ax.transAxes,
                     ha='center', fontsize=11, fontweight='bold',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    def update(frame):
        action, state, j, k, sorted_set = steps[frame]

        for i in range(n):
            bars[i].set_height(state[i])
            texts[i].set_position((i, state[i] + 1.5))
            texts[i].set_text(str(state[i]))

            if i in sorted_set:
                bars[i].set_color('#2ECC71')
            elif action in ('compare', 'swap') and i in (j, k):
                bars[i].set_color('#FFA500' if action == 'compare' else '#E74C3C')
            else:
                bars[i].set_color('steelblue')

        if action == 'compare':
            status.set_text(f'Comparando: {state[j]} vs {state[k]}')
        elif action == 'swap':
            status.set_text(f'Intercambio: {state[k]} \u2194 {state[j]}')
        elif action == 'done':
            status.set_text('¡Arreglo ordenado!')
        elif action == 'initial':
            status.set_text('Arreglo inicial')

        return list(bars) + texts + [status]

    anim = FuncAnimation(fig, update, frames=len(steps),
                         interval=400, repeat=False)
    plt.close()
    return HTML(anim.to_jshtml())
````

````python
arr = [38, 27, 43, 3, 9, 82, 10]
animate_bubble_sort(arr)
````

## Ordenamiento por Selección (*Selection Sort*)

### ¿Cómo funciona?

El Selection Sort divide el arreglo en dos partes: una **parte ordenada** (al inicio) y una **parte no ordenada** (al final).

En cada pasada:
1. Busca el **mínimo** en la parte no ordenada.
2. Lo **intercambia** con el primer elemento de la parte no ordenada.
3. La frontera entre la parte ordenada y la no ordenada avanza un lugar.

**Pseudocódigo:**

```
función selection_sort(A):
    n = tamaño(A)
    para i desde 0 hasta n-2:
        min_idx = i
        para j desde i+1 hasta n-1:
            si A[j] < A[min_idx]:
                min_idx = j
        intercambiar(A[i], A[min_idx])
```

Es como buscar la carta más pequeña de un montón y ponerla al inicio, luego buscar la siguiente más pequeña, y así sucesivamente.

### Complejidad computacional

En cada pasada $i$, se realizan $n-1-i$ comparaciones para encontrar el m\u00ednimo:

$$
\text{Total de comparaciones} = \sum_{i=0}^{n-2} (n-1-i) = \frac{n(n-1)}{2}
$$

A diferencia del Bubble Sort, el Selection Sort siempre hace exactamente $\frac{n(n-1)}{2}$ comparaciones, pero realiza a lo sumo $n-1$ intercambios (uno por pasada).

- **Mejor caso**: $O(n^2)$ (siempre recorre todo para encontrar el m\u00ednimo)
- **Peor caso**: $O(n^2)$
- **Intercambios**: $O(n)$ \u2014 mucho menos que Bubble Sort

$$
T(n) = O(n^2)
$$

### Ejemplo paso a paso

Ordenemos $[38, 27, 43, 3, 9, 82, 10]$:

| Pasada | Mínimo encontrado | Intercambio | Resultado |
|--------|---------------------|-------------|-----------|
| 1      | 3 (índice 3)       | 38 ↔ 3      | $[\mathbf{3}, 27, 43, 38, 9, 82, 10]$ |
| 2      | 9 (índice 4)       | 27 ↔ 9      | $[3, \mathbf{9}, 43, 38, 27, 82, 10]$ |
| 3      | 10 (índice 6)      | 43 ↔ 10     | $[3, 9, \mathbf{10}, 38, 27, 82, 43]$ |
| 4      | 27 (índice 4)      | 38 ↔ 27     | $[3, 9, 10, \mathbf{27}, 38, 82, 43]$ |
| 5      | 38 (índice 4)      | ya en lugar  | $[3, 9, 10, 27, \mathbf{38}, 82, 43]$ |
| 6      | 43 (índice 6)      | 82 ↔ 43     | $[3, 9, 10, 27, 38, \mathbf{43}, \mathbf{82}]$ |

````python
def selection_sort_steps(arr):
    """Registra cada paso del Selection Sort."""
    arr = list(arr)
    n = len(arr)
    steps = [('initial', list(arr), -1, -1, -1, set())]
    sorted_indices = set()

    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            steps.append(('scanning', list(arr), i, j, min_idx, set(sorted_indices)))
            if arr[j] < arr[min_idx]:
                min_idx = j

        if min_idx != i:
            steps.append(('swap', list(arr), i, min_idx, min_idx, set(sorted_indices)))
            arr[i], arr[min_idx] = arr[min_idx], arr[i]

        sorted_indices.add(i)
        steps.append(('placed', list(arr), i, -1, -1, set(sorted_indices)))

    sorted_indices.add(n - 1)
    steps.append(('done', list(arr), -1, -1, -1, set(range(n))))
    return steps


def animate_selection_sort(arr):
    """Crea una animación interactiva del Selection Sort."""
    steps = selection_sort_steps(arr)
    n = len(arr)
    max_val = max(arr)

    fig, ax = plt.subplots(figsize=(10, 5))
    bar_data = steps[0][1]
    bars = ax.bar(range(n), bar_data, color='steelblue', edgecolor='black')
    texts = []
    for i, v in enumerate(bar_data):
        t = ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
        texts.append(t)

    ax.set_xlabel('Índice')
    ax.set_ylabel('Valor')
    ax.set_title('Selection Sort')
    ax.set_xticks(range(n))
    ax.set_ylim(0, max_val * 1.25)
    ax.grid(True, axis='y')

    status = ax.text(0.5, 1.08, '', transform=ax.transAxes,
                     ha='center', fontsize=11, fontweight='bold',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    def update(frame):
        action, state, pos_i, pos_j, min_idx, sorted_set = steps[frame]

        for i in range(n):
            bars[i].set_height(state[i])
            texts[i].set_position((i, state[i] + 1.5))
            texts[i].set_text(str(state[i]))

            if i in sorted_set:
                bars[i].set_color('#2ECC71')
            elif action == 'scanning' and i == min_idx:
                bars[i].set_color('#E74C3C')  # current minimum
            elif action == 'scanning' and i == pos_j:
                bars[i].set_color('#FFA500')  # being compared
            elif action == 'scanning' and i == pos_i:
                bars[i].set_color('#F39C12')  # position to fill
            elif action == 'swap' and i in (pos_i, min_idx):
                bars[i].set_color('#E74C3C')
            else:
                bars[i].set_color('steelblue')

        if action == 'scanning':
            status.set_text(f'Buscando mínimo: comparando arr[{pos_j}]={state[pos_j]}, mínimo actual={state[min_idx]}')
        elif action == 'swap':
            status.set_text(f'Intercambio: arr[{pos_i}]={state[pos_i]} \u2194 arr[{min_idx}]={state[min_idx]}')
        elif action == 'placed':
            status.set_text(f'{state[pos_i]} colocado en posición {pos_i}')
        elif action == 'done':
            status.set_text('¡Arreglo ordenado!')
        elif action == 'initial':
            status.set_text('Arreglo inicial')

        return list(bars) + texts + [status]

    anim = FuncAnimation(fig, update, frames=len(steps),
                         interval=200, repeat=False)
    plt.close()
    return HTML(anim.to_jshtml())
````

````python
arr = [38, 27, 43, 3, 9, 82, 10]
animate_selection_sort(arr)
````

## Ordenamiento por Inserción (*Insertion Sort*)

### ¿Cómo funciona?

El Insertion Sort funciona como **ordenar cartas en la mano**: tomas una carta a la vez y la insertas en la posición correcta entre las cartas que ya están ordenadas.

El algoritmo:
1. Comienza con el segundo elemento (el primero ya está "ordenado" por sí solo).
2. Toma el elemento actual (la "clave").
3. Lo compara con los elementos anteriores y los **desplaza** hacia la derecha hasta encontrar su posición.
4. Inserta la clave en esa posición.

**Pseudocódigo:**

```
función insertion_sort(A):
    para i desde 1 hasta n-1:
        clave = A[i]
        j = i - 1
        mientras j ≥ 0 y A[j] > clave:
            A[j+1] = A[j]    (desplazar a la derecha)
            j = j - 1
        A[j+1] = clave        (insertar)
```

Es muy eficiente para arreglos **pequeños** o **casi ordenados**.

### Complejidad computacional

- **Mejor caso** (arreglo ya ordenado): cada elemento se compara una sola vez con su predecesor $\rightarrow O(n)$
- **Peor caso** (arreglo en orden inverso): cada elemento debe compararse con todos los anteriores:

$$
\text{Comparaciones} = 1 + 2 + 3 + \dots + (n-1) = \frac{n(n-1)}{2}
$$

- **Caso promedio**: $O(n^2)$

$$
T(n) = O(n^2)
$$

**Ventaja clave**: es el m\u00e1s r\u00e1pido de los algoritmos $O(n^2)$ para arreglos **casi ordenados**, ya que realiza muy pocos desplazamientos.

### Ejemplo paso a paso

Ordenemos $[38, 27, 43, 3, 9, 82, 10]$:

| Paso | Clave | Acción | Resultado |
|------|-------|--------|-----------|
| 1 | 27 | 27 < 38 → insertar antes de 38 | $[\mathbf{27, 38}, 43, 3, 9, 82, 10]$ |
| 2 | 43 | 43 > 38 → ya en lugar | $[27, 38, \mathbf{43}, 3, 9, 82, 10]$ |
| 3 | 3  | 3 < 43, 38, 27 → insertar al inicio | $[\mathbf{3}, 27, 38, 43, 9, 82, 10]$ |
| 4 | 9  | 9 < 43, 38, 27 → insertar después de 3 | $[3, \mathbf{9}, 27, 38, 43, 82, 10]$ |
| 5 | 82 | 82 > 43 → ya en lugar | $[3, 9, 27, 38, 43, \mathbf{82}, 10]$ |
| 6 | 10 | 10 < 82, 43, 38, 27 → insertar después de 9 | $[3, 9, \mathbf{10}, 27, 38, 43, 82]$ |

````python
def insertion_sort_steps(arr):
    """Registra cada paso del Insertion Sort."""
    arr = list(arr)
    n = len(arr)
    steps = [('initial', list(arr), -1, -1, set())]
    sorted_set = {0}

    for i in range(1, n):
        key = arr[i]
        steps.append(('pick_key', list(arr), i, -1, set(sorted_set)))

        j = i - 1
        while j >= 0 and arr[j] > key:
            steps.append(('shift', list(arr), j, i, set(sorted_set)))
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key
        sorted_set.add(i)
        # After insertion, indices 0..i are sorted
        steps.append(('insert', list(arr), j + 1, i, set(range(i + 1))))

    steps.append(('done', list(arr), -1, -1, set(range(n))))
    return steps


def animate_insertion_sort(arr):
    """Crea una animación interactiva del Insertion Sort."""
    steps = insertion_sort_steps(arr)
    n = len(arr)
    max_val = max(arr)

    fig, ax = plt.subplots(figsize=(10, 5))
    bar_data = steps[0][1]
    bars = ax.bar(range(n), bar_data, color='steelblue', edgecolor='black')
    texts = []
    for i, v in enumerate(bar_data):
        t = ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
        texts.append(t)

    ax.set_xlabel('Índice')
    ax.set_ylabel('Valor')
    ax.set_title('Insertion Sort')
    ax.set_xticks(range(n))
    ax.set_ylim(0, max_val * 1.25)
    ax.grid(True, axis='y')

    status = ax.text(0.5, 1.08, '', transform=ax.transAxes,
                     ha='center', fontsize=11, fontweight='bold',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    def update(frame):
        action, state, pos_a, pos_b, sorted_set = steps[frame]

        for i in range(n):
            bars[i].set_height(state[i])
            texts[i].set_position((i, state[i] + 1.5))
            texts[i].set_text(str(state[i]))

            if action == 'done' or i in sorted_set:
                bars[i].set_color('#2ECC71')
            else:
                bars[i].set_color('steelblue')

        if action == 'pick_key':
            bars[pos_a].set_color('#FFA500')
            status.set_text(f'Clave seleccionada: {state[pos_a]}')
        elif action == 'shift':
            bars[pos_a].set_color('#E74C3C')
            status.set_text(f'Desplazando {state[pos_a]} a la derecha')
        elif action == 'insert':
            bars[pos_a].set_color('#F39C12')
            for i in range(pos_b + 1):
                bars[i].set_color('#2ECC71')
            bars[pos_a].set_color('#F39C12')
            status.set_text(f'{state[pos_a]} insertado en posición {pos_a}')
        elif action == 'done':
            status.set_text('¡Arreglo ordenado!')
        elif action == 'initial':
            for i in range(n):
                bars[i].set_color('steelblue')
            status.set_text('Arreglo inicial')

        return list(bars) + texts + [status]

    anim = FuncAnimation(fig, update, frames=len(steps),
                         interval=500, repeat=False)
    plt.close()
    return HTML(anim.to_jshtml())
````

````python
arr = [38, 27, 43, 3, 9, 82, 10]
animate_insertion_sort(arr)
````

## Ordenamiento por Mezcla (*Merge Sort*)

### ¿Cómo funciona?

Merge Sort utiliza la estrategia **divide y vencerás** (*divide and conquer*):

1. **Dividir**: Divide el arreglo a la mitad, recursivamente, hasta tener subarreglos de un solo elemento.
2. **Conquistar**: Cada subarreglo de un elemento ya está ordenado.
3. **Combinar** (*merge*): Fusiona dos subarreglos ordenados en uno solo, comparando elemento por elemento.

**Pseudocódigo:**

```
función merge_sort(A, izq, der):
    si izq < der:
        mid = (izq + der) / 2
        merge_sort(A, izq, mid)
        merge_sort(A, mid+1, der)
        merge(A, izq, mid, der)

función merge(A, izq, mid, der):
    crear arreglos temporales L = A[izq..mid], R = A[mid+1..der]
    fusionar L y R en A[izq..der] comparando elemento por elemento
```

La clave del algoritmo está en la operación de **merge** (fusionar), que combina dos subarreglos ordenados en tiempo lineal.

### Complejidad computacional

La recursi\u00f3n del Merge Sort se describe con la relaci\u00f3n de recurrencia:

$$
T(n) = 2 \cdot T\left(\frac{n}{2}\right) + O(n)
$$

Donde:
- $2 \cdot T(n/2)$: ordenar las dos mitades
- $O(n)$: fusionar ambas mitades

Expandiendo la recursi\u00f3n (o aplicando el **Teorema Maestro**):

| Nivel | Subproblemas | Tama\u00f1o | Trabajo por nivel |
|-------|-------------|--------|-------------------|
| 0     | 1           | $n$    | $n$               |
| 1     | 2           | $n/2$  | $n$               |
| 2     | 4           | $n/4$  | $n$               |
| $k$   | $2^k$       | $n/2^k$| $n$               |

Hay $\log_2 n$ niveles, y cada nivel hace $O(n)$ trabajo:

$$
T(n) = n \cdot \log_2 n = O(n \log n)
$$

- **Mejor caso**: $O(n \log n)$
- **Peor caso**: $O(n \log n)$
- **Espacio adicional**: $O(n)$ (no es in-place)

$$
T(n) = O(n \log n) \quad \text{siempre}
$$

### Ejemplo paso a paso

Ordenemos $[38, 27, 43, 3, 9, 82, 10]$:

**Fase de división:**
```
[38, 27, 43, 3, 9, 82, 10]
        /                \\
[38, 27, 43, 3]     [9, 82, 10]
    /       \\          /      \\
[38, 27]  [43, 3]   [9, 82]  [10]
 /   \\    /   \\    /   \\
[38] [27] [43] [3] [9] [82]
```

**Fase de fusión (merge):**
```
[27, 38] [3, 43]   [9, 82]  [10]
    \\       /          \\      /
[3, 27, 38, 43]     [9, 10, 82]
        \\                /
[3, 9, 10, 27, 38, 43, 82]
```

````python
def merge_sort_steps(arr):
    """Registra los estados del arreglo durante Merge Sort (bottom-up para animación)."""
    arr = list(arr)
    n = len(arr)
    steps = [('initial', list(arr), -1, -1, 'Arreglo inicial')]

    def merge(arr, left, mid, right, steps):
        L = arr[left:mid + 1]
        R = arr[mid + 1:right + 1]
        i = j = 0
        k = left

        steps.append(('merging', list(arr), left, right,
                       f'Fusionando [{left}..{mid}] y [{mid+1}..{right}]'))

        while i < len(L) and j < len(R):
            if L[i] <= R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1

        steps.append(('merged', list(arr), left, right,
                       f'Resultado: {arr[left:right+1]}'))

    def sort(arr, left, right, steps):
        if left < right:
            mid = (left + right) // 2
            sort(arr, left, mid, steps)
            sort(arr, mid + 1, right, steps)
            merge(arr, left, mid, right, steps)

    sort(arr, 0, n - 1, steps)
    steps.append(('done', list(arr), 0, n - 1, '¡Arreglo ordenado!'))
    return steps


def animate_merge_sort(arr):
    """Crea una animación interactiva del Merge Sort."""
    steps = merge_sort_steps(arr)
    n = len(arr)
    max_val = max(arr)

    fig, ax = plt.subplots(figsize=(10, 5))
    bar_data = steps[0][1]
    bars = ax.bar(range(n), bar_data, color='steelblue', edgecolor='black')
    texts = []
    for i, v in enumerate(bar_data):
        t = ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
        texts.append(t)

    ax.set_xlabel('Índice')
    ax.set_ylabel('Valor')
    ax.set_title('Merge Sort')
    ax.set_xticks(range(n))
    ax.set_ylim(0, max_val * 1.25)
    ax.grid(True, axis='y')

    status = ax.text(0.5, 1.08, '', transform=ax.transAxes,
                     ha='center', fontsize=11, fontweight='bold',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    def update(frame):
        action, state, left, right, desc = steps[frame]

        for i in range(n):
            bars[i].set_height(state[i])
            texts[i].set_position((i, state[i] + 1.5))
            texts[i].set_text(str(state[i]))

            if action == 'done':
                bars[i].set_color('#2ECC71')
            elif action == 'merging' and left <= i <= right:
                bars[i].set_color('#FFA500')
            elif action == 'merged' and left <= i <= right:
                bars[i].set_color('#5DADE2')
            else:
                bars[i].set_color('steelblue')

        status.set_text(desc)
        return list(bars) + texts + [status]

    anim = FuncAnimation(fig, update, frames=len(steps),
                         interval=800, repeat=False)
    plt.close()
    return HTML(anim.to_jshtml())
````

## Ordenamiento Rápido (*Quick Sort*)

### ¿Cómo funciona?

Quick Sort también usa **divide y vencerás**, pero de forma diferente al Merge Sort:

1. **Elegir un pivote**: se selecciona un elemento del arreglo (usualmente el último).
2. **Particionar**: reorganizar el arreglo de modo que todos los elementos **menores** al pivote queden a su izquierda, y todos los **mayores** a su derecha.
3. **Recursión**: aplicar Quick Sort a la parte izquierda y a la parte derecha.

**Pseudocódigo:**

```
función quick_sort(A, low, high):
    si low < high:
        p = particionar(A, low, high)
        quick_sort(A, low, p - 1)
        quick_sort(A, p + 1, high)

función particionar(A, low, high):
    pivote = A[high]
    i = low - 1
    para j desde low hasta high-1:
        si A[j] ≤ pivote:
            i += 1
            intercambiar(A[i], A[j])
    intercambiar(A[i+1], A[high])
    retornar i + 1
```

Después de la partición, el pivote queda en su **posición final definitiva**.

### Complejidad computacional

**Caso promedio:** Si el pivote divide el arreglo aproximadamente a la mitad cada vez:

$$
T(n) = 2 \cdot T\left(\frac{n}{2}\right) + O(n) = O(n \log n)
$$

**Peor caso:** Si el pivote es siempre el m\u00e1ximo o m\u00ednimo (arreglo ya ordenado con pivote al final):

$$
T(n) = T(n-1) + O(n) = O(n^2)
$$

**Resumen:**

| Caso | Complejidad |
|------|------------|
| Mejor | $O(n \log n)$ |
| Promedio | $O(n \log n)$ |
| Peor | $O(n^2)$ |

A pesar de su peor caso $O(n^2)$, Quick Sort es en la pr\u00e1ctica uno de los algoritmos m\u00e1s r\u00e1pidos debido a su excelente uso de la cach\u00e9 de memoria y su constante multiplicativa peque\u00f1a.

### Ejemplo paso a paso

Ordenemos $[38, 27, 43, 3, 9, 82, 10]$, usando el último elemento como pivote:

**Paso 1: Particionar con pivote = 10**

Recorremos el arreglo y movemos los menores a 10 a la izquierda:
- 38 > 10 → no mover
- 27 > 10 → no mover
- 43 > 10 → no mover
- 3 < 10 → mover (intercambiar con posición 0)
- 9 < 10 → mover (intercambiar con posición 1)

Colocar pivote: $[3, 9, \mathbf{10}, 38, 43, 82, 27]$

El **10** está en su posición final (2). Ahora aplicar Quick Sort a $[3, 9]$ (izquierda) y $[38, 43, 82, 27]$ (derecha).

El proceso continúa recursivamente hasta que todo esté ordenado.

````python
def quick_sort_steps(arr):
    """Registra cada paso del Quick Sort."""
    arr = list(arr)
    n = len(arr)
    steps = [('initial', list(arr), -1, -1, -1, 'Arreglo inicial')]
    final_positions = set()

    def partition(arr, low, high, steps):
        pivot = arr[high]
        steps.append(('pivot', list(arr), low, high, high,
                       f'Pivote = {pivot} (\u00edndice {high})'))
        i = low - 1

        for j in range(low, high):
            steps.append(('compare', list(arr), j, high, i + 1,
                           f'arr[{j}]={arr[j]} vs pivote={pivot}'))
            if arr[j] <= pivot:
                i += 1
                if i != j:
                    arr[i], arr[j] = arr[j], arr[i]
                    steps.append(('swap', list(arr), i, j, -1,
                                   f'Intercambio arr[{i}] \u2194 arr[{j}]'))

        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        steps.append(('place_pivot', list(arr), i + 1, low, high,
                       f'Pivote {pivot} colocado en posición {i+1}'))
        final_positions.add(i + 1)
        return i + 1

    def qsort(arr, low, high, steps):
        if low < high:
            p = partition(arr, low, high, steps)
            qsort(arr, low, p - 1, steps)
            qsort(arr, p + 1, high, steps)
        elif low == high:
            final_positions.add(low)

    qsort(arr, 0, n - 1, steps)
    steps.append(('done', list(arr), -1, -1, -1, '¡Arreglo ordenado!'))
    return steps, final_positions


def animate_quick_sort(arr):
    """Crea una animación interactiva del Quick Sort."""
    steps, final_pos = quick_sort_steps(arr)
    n = len(arr)
    max_val = max(arr)

    fig, ax = plt.subplots(figsize=(10, 5))
    bar_data = steps[0][1]
    bars = ax.bar(range(n), bar_data, color='steelblue', edgecolor='black')
    texts = []
    for i, v in enumerate(bar_data):
        t = ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
        texts.append(t)

    ax.set_xlabel('Índice')
    ax.set_ylabel('Valor')
    ax.set_title('Quick Sort')
    ax.set_xticks(range(n))
    ax.set_ylim(0, max_val * 1.25)
    ax.grid(True, axis='y')

    status = ax.text(0.5, 1.08, '', transform=ax.transAxes,
                     ha='center', fontsize=11, fontweight='bold',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    # Track which pivots are placed
    placed_pivots = set()

    def update(frame):
        nonlocal placed_pivots
        action, state, a, b, c, desc = steps[frame]

        # Rebuild placed pivots up to this frame
        placed_pivots = set()
        for f in range(frame + 1):
            if steps[f][0] == 'place_pivot':
                placed_pivots.add(steps[f][2])

        for i in range(n):
            bars[i].set_height(state[i])
            texts[i].set_position((i, state[i] + 1.5))
            texts[i].set_text(str(state[i]))

            if action == 'done':
                bars[i].set_color('#2ECC71')
            elif i in placed_pivots:
                bars[i].set_color('#2ECC71')
            elif action == 'pivot' and i == c:
                bars[i].set_color('#9B59B6')  # purple for pivot
            elif action == 'compare' and i == a:
                bars[i].set_color('#FFA500')
            elif action == 'compare' and i == b:
                bars[i].set_color('#9B59B6')  # pivot
            elif action == 'swap' and i in (a, b):
                bars[i].set_color('#E74C3C')
            elif action == 'place_pivot' and i == a:
                bars[i].set_color('#2ECC71')
            else:
                bars[i].set_color('steelblue')

        status.set_text(desc)
        return list(bars) + texts + [status]

    anim = FuncAnimation(fig, update, frames=len(steps),
                         interval=500, repeat=False)
    plt.close()
    return HTML(anim.to_jshtml())
````

## Comparación de Algoritmos

Resumamos las diferencias clave:

| Algoritmo | Mejor | Promedio | Peor | Espacio | Estable | In-place |
|---|---|---|---|---|---|---|
| **Bubble Sort** | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ | S\u00ed | S\u00ed |
| **Selection Sort** | $O(n^2)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ | No | S\u00ed |
| **Insertion Sort** | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ | S\u00ed | S\u00ed |
| **Merge Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n \log n)$ | $O(n)$ | S\u00ed | No |
| **Quick Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n^2)$ | $O(\log n)$ | No | S\u00ed |

**\u00bfCu\u00e1l usar?**
- **Pocos elementos** o **casi ordenado**: Insertion Sort
- **Garant\u00eda de rendimiento**: Merge Sort (siempre $O(n \log n)$)
- **Mejor rendimiento pr\u00e1ctico**: Quick Sort (r\u00e1pido en la mayor\u00eda de casos)

````python
import numpy as np
import matplotlib.pyplot as plt

n_values = np.arange(1, 501)
quadratic = n_values ** 2
nlogn = n_values * np.log2(np.maximum(n_values, 1))
linear = n_values.astype(float)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico 1: Crecimiento teórico
axes[0].plot(n_values, quadratic, label='$O(n^2)$ — Bubble/Selection/Insertion', color='#E74C3C', linewidth=2)
axes[0].plot(n_values, nlogn, label='$O(n \log n)$ — Merge/Quick', color='#2ECC71', linewidth=2)
axes[0].plot(n_values, linear, label='$O(n)$ — Mejor caso Bubble/Insertion', color='#3498DB', linewidth=2, linestyle='--')
axes[0].set_xlabel('Tamaño del arreglo ($n$)')
axes[0].set_ylabel('Operaciones')
axes[0].set_title('Crecimiento teórico de complejidad')
axes[0].legend(fontsize=9)
axes[0].grid(True)
axes[0].set_ylim(0, 50000)

# Gráfico 2: Medición real con nuestros datos
import time

def measure_sort(sort_fn, arr, runs=100):
    total = 0
    for _ in range(runs):
        a = list(arr)
        start = time.perf_counter()
        sort_fn(a)
        total += time.perf_counter() - start
    return total / runs * 1e6  # microseconds

def _bubble(a):
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]

def _selection(a):
    n = len(a)
    for i in range(n - 1):
        m = i
        for j in range(i + 1, n):
            if a[j] < a[m]:
                m = j
        a[i], a[m] = a[m], a[i]

def _insertion(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key

def _merge_sort(a):
    if len(a) > 1:
        mid = len(a) // 2
        L, R = a[:mid], a[mid:]
        _merge_sort(L)
        _merge_sort(R)
        i = j = k = 0
        while i < len(L) and j < len(R):
            if L[i] <= R[j]:
                a[k] = L[i]; i += 1
            else:
                a[k] = R[j]; j += 1
            k += 1
        while i < len(L):
            a[k] = L[i]; i += 1; k += 1
        while j < len(R):
            a[k] = R[j]; j += 1; k += 1

def _quick_sort(a, lo=None, hi=None):
    if lo is None: lo = 0
    if hi is None: hi = len(a) - 1
    if lo < hi:
        pivot = a[hi]; i = lo - 1
        for j in range(lo, hi):
            if a[j] <= pivot:
                i += 1; a[i], a[j] = a[j], a[i]
        a[i+1], a[hi] = a[hi], a[i+1]
        _quick_sort(a, lo, i); _quick_sort(a, i+2, hi)

test_arr = list(np.random.randint(0, 1000, 200))
algos = ['Bubble', 'Selection', 'Insertion', 'Merge', 'Quick']
times = [
    measure_sort(_bubble, test_arr),
    measure_sort(_selection, test_arr),
    measure_sort(_insertion, test_arr),
    measure_sort(_merge_sort, test_arr),
    measure_sort(lambda a: _quick_sort(a), test_arr),
]
colors = ['#E74C3C', '#E67E22', '#F1C40F', '#2ECC71', '#3498DB']

bars = axes[1].bar(algos, times, color=colors, edgecolor='black')
for bar, t in zip(bars, times):
    axes[1].text(bar.get_x() + bar.get_width() / 2, bar.get_height() + max(times) * 0.02,
                 f'{t:.0f}µs', ha='center', fontweight='bold', fontsize=9)
axes[1].set_ylabel('Tiempo (µs)')
axes[1].set_title(f'Tiempo real promedio (n={len(test_arr)}, 100 ejecuciones)')
axes[1].grid(True, axis='y')

plt.tight_layout()
plt.show()
````

## Implementación con Código

### Funciones

````python
def bubble_sort(arr):
    """Bubble Sort: compara e intercambia pares adyacentes."""
    arr = list(arr)
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break  # optimización: si no hubo intercambios, ya está ordenado
    return arr


def selection_sort(arr):
    """Selection Sort: selecciona el mínimo en cada pasada."""
    arr = list(arr)
    n = len(arr)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


def insertion_sort(arr):
    """Insertion Sort: inserta cada elemento en su posición correcta."""
    arr = list(arr)
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr


def merge_sort(arr):
    """Merge Sort: divide y vencerás con fusión."""
    arr = list(arr)
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    # Fusionar
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result


def quick_sort(arr):
    """Quick Sort: partición con pivote."""
    arr = list(arr)
    def _qsort(a, lo, hi):
        if lo < hi:
            pivot = a[hi]
            i = lo - 1
            for j in range(lo, hi):
                if a[j] <= pivot:
                    i += 1
                    a[i], a[j] = a[j], a[i]
            a[i + 1], a[hi] = a[hi], a[i + 1]
            p = i + 1
            _qsort(a, lo, p - 1)
            _qsort(a, p + 1, hi)
    _qsort(arr, 0, len(arr) - 1)
    return arr
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


class SortVisualizer:
    """Clase unificada para ejecutar y visualizar algoritmos de ordenamiento."""

    def __init__(self, arr):
        self.original = list(arr)
        self.arr = list(arr)
        self.states = [list(arr)]  # lista de estados del arreglo
        self.algorithm = None

    def _record(self, arr):
        self.states.append(list(arr))

    def bubble_sort(self):
        """Bubble Sort."""
        self.arr = list(self.original)
        self.algorithm = 'Bubble Sort'
        self.states = [list(self.arr)]
        n = len(self.arr)
        for i in range(n - 1):
            for j in range(n - 1 - i):
                if self.arr[j] > self.arr[j + 1]:
                    self.arr[j], self.arr[j + 1] = self.arr[j + 1], self.arr[j]
                    self._record(self.arr)
        return list(self.arr)

    def selection_sort(self):
        """Selection Sort."""
        self.arr = list(self.original)
        self.algorithm = 'Selection Sort'
        self.states = [list(self.arr)]
        n = len(self.arr)
        for i in range(n - 1):
            min_idx = i
            for j in range(i + 1, n):
                if self.arr[j] < self.arr[min_idx]:
                    min_idx = j
            if min_idx != i:
                self.arr[i], self.arr[min_idx] = self.arr[min_idx], self.arr[i]
            self._record(self.arr)
        return list(self.arr)

    def insertion_sort(self):
        """Insertion Sort."""
        self.arr = list(self.original)
        self.algorithm = 'Insertion Sort'
        self.states = [list(self.arr)]
        for i in range(1, len(self.arr)):
            key = self.arr[i]
            j = i - 1
            while j >= 0 and self.arr[j] > key:
                self.arr[j + 1] = self.arr[j]
                j -= 1
            self.arr[j + 1] = key
            self._record(self.arr)
        return list(self.arr)

    def merge_sort(self):
        """Merge Sort."""
        self.arr = list(self.original)
        self.algorithm = 'Merge Sort'
        self.states = [list(self.arr)]
        def _merge_sort(arr, l, r):
            if l < r:
                m = (l + r) // 2
                _merge_sort(arr, l, m)
                _merge_sort(arr, m + 1, r)
                L, R = arr[l:m+1], arr[m+1:r+1]
                i = j = 0; k = l
                while i < len(L) and j < len(R):
                    if L[i] <= R[j]: arr[k] = L[i]; i += 1
                    else: arr[k] = R[j]; j += 1
                    k += 1
                while i < len(L): arr[k] = L[i]; i += 1; k += 1
                while j < len(R): arr[k] = R[j]; j += 1; k += 1
                self._record(arr)
        _merge_sort(self.arr, 0, len(self.arr) - 1)
        return list(self.arr)

    def quick_sort(self):
        """Quick Sort."""
        self.arr = list(self.original)
        self.algorithm = 'Quick Sort'
        self.states = [list(self.arr)]
        def _qsort(arr, lo, hi):
            if lo < hi:
                pivot = arr[hi]; i = lo - 1
                for j in range(lo, hi):
                    if arr[j] <= pivot:
                        i += 1; arr[i], arr[j] = arr[j], arr[i]
                arr[i+1], arr[hi] = arr[hi], arr[i+1]
                self._record(arr)
                p = i + 1
                _qsort(arr, lo, p - 1)
                _qsort(arr, p + 1, hi)
        _qsort(self.arr, 0, len(self.arr) - 1)
        return list(self.arr)

    def plot_progress(self):
        """Muestra el progreso del ordenamiento en subplots."""
        if len(self.states) <= 1:
            raise Exception("Primero ejecuta un algoritmo de ordenamiento.")

        n_show = min(6, len(self.states))
        indices = np.linspace(0, len(self.states) - 1, n_show, dtype=int)
        n = len(self.original)
        max_val = max(self.original)
        sorted_arr = sorted(self.original)

        fig, axes = plt.subplots(1, n_show, figsize=(3.5 * n_show, 4), sharey=True)
        if n_show == 1:
            axes = [axes]

        for ax_idx, state_idx in enumerate(indices):
            state = self.states[state_idx]
            is_sorted = (state == sorted_arr)
            colors = ['#2ECC71' if is_sorted else 'steelblue'] * n
            axes[ax_idx].bar(range(n), state, color=colors, edgecolor='black')
            for i, v in enumerate(state):
                axes[ax_idx].text(i, v + 1, str(v), ha='center', fontsize=7, fontweight='bold')
            label = 'Inicial' if state_idx == 0 else ('Final' if state_idx == len(self.states) - 1 else f'Paso {state_idx}')
            axes[ax_idx].set_title(label, fontsize=9)
            axes[ax_idx].set_xticks(range(n))
            axes[ax_idx].grid(True, axis='y')

        axes[0].set_ylabel('Valor')
        plt.suptitle(f'{self.algorithm} — Progreso ({len(self.states)} estados)',
                     fontsize=12, fontweight='bold')
        plt.tight_layout()
        plt.show()

    def animate(self):
        """Retorna una animación interactiva del ordenamiento."""
        if len(self.states) <= 1:
            raise Exception("Primero ejecuta un algoritmo de ordenamiento.")

        n = len(self.original)
        max_val = max(self.original)
        sorted_arr = sorted(self.original)

        fig, ax = plt.subplots(figsize=(10, 5))
        bar_data = self.states[0]
        bars = ax.bar(range(n), bar_data, color='steelblue', edgecolor='black')
        texts = []
        for i, v in enumerate(bar_data):
            t = ax.text(i, v + 1.5, str(v), ha='center', fontweight='bold')
            texts.append(t)

        ax.set_xlabel('Índice')
        ax.set_ylabel('Valor')
        ax.set_title(self.algorithm)
        ax.set_xticks(range(n))
        ax.set_ylim(0, max_val * 1.25)
        ax.grid(True, axis='y')

        status = ax.text(0.5, 1.08, '', transform=ax.transAxes,
                         ha='center', fontsize=11, fontweight='bold',
                         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

        states = self.states

        def update(frame):
            state = states[frame]
            done = (state == sorted_arr)
            for i in range(n):
                bars[i].set_height(state[i])
                texts[i].set_position((i, state[i] + 1.5))
                texts[i].set_text(str(state[i]))
                bars[i].set_color('#2ECC71' if done else 'steelblue')

            if done:
                status.set_text('¡Ordenado!')
            else:
                status.set_text(f'Paso {frame + 1} / {len(states)}')
            return list(bars) + texts + [status]

        anim = FuncAnimation(fig, update, frames=len(states),
                             interval=400, repeat=False)
        plt.close()
        return HTML(anim.to_jshtml())
````

````python
# Ejemplo con la clase
arr = [38, 27, 43, 3, 9, 82, 10]

vis = SortVisualizer(arr)
result = vis.merge_sort()
print(f"Merge Sort: {result}")
vis.plot_progress()
````

````python
# Animación con la clase
vis2 = SortVisualizer([38, 27, 43, 3, 9, 82, 10])
vis2.quick_sort()
vis2.animate()
````

````python
# Comparar todos los algoritmos
arr = [38, 27, 43, 3, 9, 82, 10]
algorithms = ['bubble_sort', 'selection_sort', 'insertion_sort', 'merge_sort', 'quick_sort']
names = ['Bubble', 'Selection', 'Insertion', 'Merge', 'Quick']

print(f"Arreglo original: {arr}")
print(f"{'='*50}")
for algo, name in zip(algorithms, names):
    vis = SortVisualizer(arr)
    result = getattr(vis, algo)()
    print(f"{name:12s} \u2192 {result}  ({len(vis.states)} estados)")
````
